TUPLES

```python
Tuples are compound data types. They are expressed as comma-separated elements inside parentheses.
Example: Consider a tuple Cars as shown below:
```

The different types in Python are Int, Float, and String.

The tuple Cars contains different types of data.

Each element of the tuple can be accessed by an index.

Example:

Cars[0] = "I20"

Cars[1] = "Baleno"

and so on.

Negative indexing can also be used.

Example:

Cars[-1] = 7.5

Cars[-2] = 2019

Cars[-6] = "I20"





```python
Cars = ('I20','Baleno','Swift','Nexon',2019,7.5)
```

```python
type(Cars)
```

```python
Cars[0]
```

```python
Cars[5]
```

```python
Cars[-1] #Access to the last element of the tuple using -ve indexing
```

```python
CarsAdd = ('Hyundai','Maruti','ECO',77)
```

```python
CarsNew = Cars + CarsAdd    # Cars tuple concatenated with CarsAdd1 tuple to create a new tuple CarsNew
print(CarsNew)
```

**L**ists

Lists are compound data types. They are expressed as comma-separated elements inside square brackets.


Example: Consider a list Cars as shown below

```python
Cars = ['I20', 'Baleno', 'Swift', 'Nexon', 2019, 7.5]
```

The different types in Python are Int, Float, and String.

The list Cars contains different types of data.
Each element of the list can be accessed by an index.

Example:
Cars[0] = "I20"
Cars[1] = "Baleno"
and so on.

Negative indexing can also be used.
Example:
Cars[-1] = 7.5
Cars[-2] = 2019
Cars[-6] = "I20"

```python
Cars = ['I20', 'Baleno', 'Swift', 'Nexon', 2019, 7.5]   # create a list Cars
```
